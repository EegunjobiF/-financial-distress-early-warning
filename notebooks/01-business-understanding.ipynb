```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns 

from scipy.io import arff

# Define file location
data_path = "../data/raw/1year.arff"

# Load ARFF file
raw_data, metadata = arff.loadarff(data_path)

df = pd.DataFrame(raw_data)

# Decode byte-string columns if present
for col in df.select_dtypes(include=["object"]).columns:
    df[col] = df[col].map(
        lambda value: (
            value.decode("utf-8")
            if isinstance(value, bytes)
            else value
        )
    )

print("Dataset shape:", df.shape)
print("First five rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()
```

```python
# Show ALL rows - no hiding
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# Pre-compute once - avoid calling isna() twice
is_na = df.isna()

# 1. Missing values - one pass, vectorized
missing_summary = (
    pd.DataFrame({
        "missing_count": is_na.sum(),
        "missing_pct": is_na.mean() * 100
    })
    .query("missing_pct > 0")
    .sort_values("missing_pct", ascending=False)
)

print("Columns with missing values:", len(missing_summary))
display(missing_summary)  # now shows all columns with missing, not just head(20)

# 2. Duplicate rows
print(f"\nDuplicate rows: {df.duplicated().sum()} ({df.duplicated().mean()*100:.2f}%)")

# 3. Unique values - check for constant columns
nunique = df.nunique(dropna=False)
constant_cols = nunique[nunique <= 1]

print(f"\nConstant columns (no predictive value): {len(constant_cols)}")
if not constant_cols.empty:
    print(constant_cols)
else:
    print("None - all columns have variance")

# 4. Numeric summary - more informative - FULL 64 ROWS
print("\nNumeric Summary (All 64 Attrs):")
summary = df.describe(percentiles=[0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99])
display(
    summary.T[['count', 'mean', 'std', 'min', '1%', '5%', '50%', '95%', '99%', 'max']]
)

# 5. Infinite values - optimized check
numeric_cols = df.select_dtypes(include="number").columns
inf_count = np.isinf(df[numeric_cols].to_numpy()).sum()

print(f"\nInfinite values: {inf_count}")
if inf_count > 0:
    # Show which columns have inf
    inf_cols = df[numeric_cols].columns[np.isinf(df[numeric_cols]).any()]
    print("Columns with inf:", inf_cols.tolist())
```

```python

# ============================================================
# CELL 3: MISSINGNESS CORRELATION & BANKRUPTCY RELATIONSHIP
# ============================================================
sns.set_theme(style="whitegrid", context="notebook")

# ------------------------------------------------------------
# PREPARATION
# ------------------------------------------------------------

# Select the original financial-ratio columns.
attr_cols = [
    col for col in df.columns
    if col.startswith("Attr")
    and col[len("Attr"):].isdigit()
]

# Create Boolean indicators: True = missing, False = present.
missing_flags = df[attr_cols].isna()

# Convert target to numeric.
target = pd.to_numeric(
    df["class"].astype(str).str.strip(),
    errors="coerce"
)

# Keep only observations with valid target labels.
valid_target = target.isin([0, 1])

# ============================================================
# 1. CORRELATION HEATMAP: DO MISSING COLUMNS MISS TOGETHER?
# ============================================================

# Focus on columns with at least one missing value.
missing_pct = missing_flags.mean() * 100

cols_with_missing = missing_pct[
    missing_pct > 0
].sort_values(ascending=False).index.tolist()

# Correlate the missingness indicators, not the original values.
missing_corr = (
    missing_flags[cols_with_missing]
    .astype(int)
    .corr()
)

# Exclude constant indicators with undefined correlations.
valid_corr_cols = missing_corr.columns[
    missing_corr.notna().all(axis=0)
].tolist()

missing_corr = missing_corr.loc[
    valid_corr_cols, valid_corr_cols
]

fig, ax = plt.subplots(figsize=(16, 13))

sns.heatmap(
    missing_corr,
    cmap="vlag",
    vmin=-1,
    vmax=1,
    center=0,
    annot=len(valid_corr_cols) <= 20,
    fmt=".2f",
    annot_kws={"size": 8},
    linewidths=0.3,
    square=True,
    cbar_kws={
        "label": "Correlation of missingness",
        "shrink": 0.75
    },
    ax=ax
)

ax.set_title(
    "Do Missing Financial Ratios Occur Together?",
    fontsize=17,
    fontweight="bold",
    pad=20
)
ax.set_xlabel("Financial Ratio Missingness", fontsize=12)
ax.set_ylabel("Financial Ratio Missingness", fontsize=12)

ax.tick_params(axis="x", labelrotation=45)
ax.tick_params(axis="y", labelrotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 2. GROUPED BAR CHART: IS MISSINGNESS RELATED TO BANKRUPTCY?
# ============================================================

# Use columns with meaningful missingness for readability.
# This includes the 2 main columns and other columns with
# at least 1% missing values.
focus_cols = missing_pct[
    missing_pct >= 1
].index.tolist()

rate_records = []

for col in focus_cols:

    is_missing = df[col].isna()

    for missing_status, group_mask in [
        ("Present", ~is_missing),
        ("Missing", is_missing)
    ]:

        mask = valid_target & group_mask
        group_target = target.loc[mask]

        if len(group_target) == 0:
            continue

        rate_records.append({
            "Attribute": col,
            "Missing Status": missing_status,
            "Bankruptcy Rate (%)": group_target.mean() * 100,
            "Company Count": len(group_target)
        })

rate_df = pd.DataFrame(rate_records)

fig, ax = plt.subplots(figsize=(14, 7))

sns.barplot(
    data=rate_df,
    x="Attribute",
    y="Bankruptcy Rate (%)",
    hue="Missing Status",
    hue_order=["Present", "Missing"],
    palette={
        "Present": "#2878B5",
        "Missing": "#D9485F"
    },
    errorbar=None,
    ax=ax
)

# Label each bar with its bankruptcy rate.
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=3,
        fontsize=9
    )

ax.set_title(
    "Bankruptcy Rate: Missing vs Present Financial Ratios",
    fontsize=17,
    fontweight="bold",
    pad=20
)
ax.set_xlabel("Financial Ratio", fontsize=12)
ax.set_ylabel("Bankruptcy Rate (%)", fontsize=12)

ax.legend(
    title="Financial Ratio Status",
    frameon=False
)

ax.tick_params(axis="x", labelrotation=0)
ax.set_ylim(
    0,
    max(1, rate_df["Bankruptcy Rate (%)"].max() * 1.25)
)

sns.despine()
plt.tight_layout()
plt.show()


# Print the supporting numbers, including sample sizes.
print("\nBankruptcy rate and sample size by missingness status:")

display(
    rate_df.sort_values(
        ["Attribute", "Missing Status"]
    ).round({
        "Bankruptcy Rate (%)": 2
    })
)
```

```python

# General plotting settings
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

TARGET = "class"

# Check that the target column exists
if TARGET not in df.columns:
    raise ValueError(
        f"Target column '{TARGET}' not found. "
        f"Available columns: {df.columns.tolist()}"
    )

# Select numerical features
numeric_features = [
    col for col in df.select_dtypes(include="number").columns
    if col != TARGET
]

# ============================================================
# STEP 3: TARGET VARIABLE ANALYSIS
# ============================================================

print("\n" + "=" * 60)
print("STEP 3: TARGET VARIABLE ANALYSIS")
print("=" * 60)

# Counts and percentages
target_summary = pd.DataFrame({
    "Company_Count": df[TARGET].value_counts(),
    "Percentage": df[TARGET].value_counts(normalize=True) * 100
}).sort_index()

print("\nBankruptcy distribution:")
print(target_summary.round(2))

# Visualize target distribution
plt.figure(figsize=(8, 5))
ax = sns.countplot(data=df, x=TARGET)

total = len(df)

for bar in ax.patches:
    count = int(bar.get_height())
    percentage = count / total * 100
    ax.annotate(
        f"{count:,}\n({percentage:.2f}%)",
        (bar.get_x() + bar.get_width() / 2, bar.get_height()),
        ha="center",
        va="bottom"
    )

plt.title("Distribution of Bankruptcy Status")
plt.xlabel("Bankruptcy Status (0 = Non-bankrupt, 1 = Bankrupt)")
plt.ylabel("Number of Companies")
plt.tight_layout()
plt.show()

# ============================================================
# STEP 4: DISTRIBUTIONS AND OUTLIERS
# ============================================================

print("\n" + "=" * 60)
print("STEP 4: DISTRIBUTIONS AND OUTLIERS")
print("=" * 60)

# Descriptive statistics
print("\nDescriptive statistics for numerical features:")
display(df[numeric_features].describe().T.round(3))

# Missing values and unique values
feature_summary = pd.DataFrame({
    "Missing_Count": df[numeric_features].isna().sum(),
    "Missing_Percent": df[numeric_features].isna().mean() * 100,
    "Unique_Values": df[numeric_features].nunique()
}).sort_values("Missing_Percent", ascending=False)

print("\nMissingness and uniqueness:")
display(feature_summary.round(2))

# Plot distributions in batches to avoid an enormous figure.
# First inspect the first 12 numeric features.
features_to_plot = numeric_features[:12]

if features_to_plot:
    df[features_to_plot].hist(
        bins=30,
        figsize=(16, 12),
        edgecolor="black"
    )
    plt.suptitle(
        "Distributions of Selected Financial Attributes",
        y=1.02
    )
    plt.tight_layout()
    plt.show()

# Boxplots to inspect extreme values
if features_to_plot:
    ncols = 3
    nrows = int(np.ceil(len(features_to_plot) / ncols))

    fig, axes = plt.subplots(
        nrows, ncols,
        figsize=(16, 4 * nrows)
    )
    axes = np.array(axes).reshape(-1)

    for ax, col in zip(axes, features_to_plot):
        sns.boxplot(x=df[col], ax=ax)
        ax.set_title(f"Outliers: {col}")
        ax.set_xlabel(col)

    for ax in axes[len(features_to_plot):]:
        ax.set_visible(False)

    plt.tight_layout()
    plt.show()

# IQR-based outlier counts
outlier_rows = []

for col in numeric_features:
    series = df[col].dropna()

    if series.empty:
        continue

    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1

    if iqr == 0:
        lower_bound = q1
        upper_bound = q3
    else:
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

    outlier_count = (
        (series < lower_bound) | (series > upper_bound)
    ).sum()

    outlier_rows.append({
        "Attribute": col,
        "Outlier_Count": int(outlier_count),
        "Outlier_Percent": outlier_count / len(series) * 100,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound
    })

outlier_summary = pd.DataFrame(outlier_rows)

if not outlier_summary.empty:
    outlier_summary = outlier_summary.sort_values(
        "Outlier_Percent", ascending=False
    )

print("\nIQR-based outlier summary:")
display(outlier_summary.round(3))

# ============================================================
# STEP 5: CORRELATION AND REDUNDANT FEATURES
# ============================================================

print("\n" + "=" * 60)
print("STEP 5: CORRELATION AND REDUNDANT FEATURES")
print("=" * 60)

# Correlation between numerical features
correlation_matrix = df[numeric_features].corr()

# Full correlation heatmap
plt.figure(
    figsize=(
        max(16, len(numeric_features) * 0.35),
        max(12, len(numeric_features) * 0.30)
    )
)

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    annot=False,
    square=True,
    linewidths=0.1,
    cbar_kws={"label": "Pearson Correlation"}
)

plt.title("Correlation Heatmap of Financial Attributes")
plt.xticks(rotation=90, fontsize=7)
plt.yticks(rotation=0, fontsize=7)
plt.tight_layout()
plt.show()

# Identify highly correlated pairs without duplicate pairs
threshold = 0.90

upper_triangle = correlation_matrix.where(
    np.triu(
        np.ones(correlation_matrix.shape),
        k=1
    ).astype(bool)
)

high_corr_pairs = (
    upper_triangle
    .stack()
    .reset_index()
)

high_corr_pairs.columns = [
    "Attribute_1",
    "Attribute_2",
    "Correlation"
]

high_corr_pairs["Absolute_Correlation"] = (
    high_corr_pairs["Correlation"].abs()
)

high_corr_pairs = high_corr_pairs[
    high_corr_pairs["Absolute_Correlation"] >= threshold
].sort_values(
    "Absolute_Correlation",
    ascending=False
)

print(f"\nHighly correlated pairs (|r| >= {threshold}):")

if high_corr_pairs.empty:
    print("No pairs met the selected correlation threshold.")
else:
    display(high_corr_pairs.round(4))

# Correlation of each feature with bankruptcy status
# Useful for exploratory analysis, not proof of predictive value.
target_correlations = (
    df[numeric_features + [TARGET]]
    .corr(numeric_only=True)[TARGET]
    .drop(TARGET)
    .dropna()
    .sort_values(key=abs, ascending=False)
)

print("\nFeatures most correlated with bankruptcy status:")
display(target_correlations.head(20).to_frame(
    name="Correlation_with_Bankruptcy"
).round(4))

print("\nEDA steps 3, 4 and 5 completed.")
```
